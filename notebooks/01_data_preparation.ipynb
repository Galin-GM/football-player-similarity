Player Similarity Data Preparation

```python
import pandas as pd

df = pd.read_csv('../data/raw/standard-stats.csv')

players_df = df[['Player', 'Squad', 'Pos', 'Min', '90s', 'Gls90', 'Ast90']]

eligible_players_df = players_df.loc[players_df['Min'] >= 900].copy()

eligible_players_df.describe()

eligible_players_df.to_csv('../data/processed/player_contributions.csv', index=False)
```
