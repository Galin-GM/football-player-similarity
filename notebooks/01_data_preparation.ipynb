# Player Similarity - Data Preparation

This notebook prepares the football player data for clustering using goals and assists per 90 minutes.
The cleaned output is saved for later analysis.

## Loading the raw data

The data source is fbref premier league 25/26 stats. The original data remains unchanged so the preparation is reproducible.

Source: https://fbref.com/en/comps/9/2025-2026/stats/2025-2026-Premier-League-Stats

```python
import pandas as pd

df = pd.read_csv('../data/raw/standard-stats.csv')
```

## Selecting relevant columns

The first version of the model uses goals and assists per 90 minutes as its clustering features. Later further stats will be included such as passes, cards, etc.

```python
players_df = df[['Player', 'Squad', 'Pos', 'Min', '90s', 'Gls90', 'Ast90']]
```

## Applying minimum playing time

Only players with at least 900 minutes are retained. This reduces the influence of unstable per 90 mins statistics produced by small playing time samples.

```python
eligible_players_df = players_df.loc[players_df['Min'] >= 900].copy()
```

## Creating and assigning primary position column

Some players have multiple positions listed such as defender and midfielder. For initial model only take first listed position as primary position.

```python
eligible_players_df['PrimaryPos'] = (
    eligible_players_df['Pos'].str[:2]
)
```

## Saving processed data

The processed data is saved as a seperate dataset for exploration and modelling.

```python
eligible_players_df.to_csv('../data/processed/player_contributions.csv', index=False)
```
