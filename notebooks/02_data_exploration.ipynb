## Player Similarity - Data Exploration

This notebooks explores the processed player data before clustering.

```python
import pandas as pd

df = pd.read_csv('../data/processed/player_contributions.csv')
```

## Relationship between goals and assists

Each point represents an player. The plot shows how players are distributed according to goals and assists per 90 minutes.

```python
df.plot.scatter(title='Goals and Assists per 90 Minutes', 
                x='Gls90', xlabel='Goals per 90 minutes', 
                y='Ast90', ylabel='Assists per 90 minutes',
                alpha=0.6)
```

```python
df.sort_values('Gls90', ascending=False)
```

```python
df.value_counts('PrimaryPos')
```

```python
df.groupby('PrimaryPos')[['Gls90', 'Ast90']].agg(['count', 'mean', 'max']).round(3)
```
